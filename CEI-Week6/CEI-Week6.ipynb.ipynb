```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
```

# **Creating spark session**

```python
spark = SparkSession.builder\
.appName("Week6_Assignment")\
.getOrCreate()
```

# **Reading CSV file**

```python
df = spark.read \
    .option("header", "true") \
    .option("inferSchema", "false") \
    .csv("Sample - Superstore.csv")
```

```python
df.printSchema()
```

```python
df.show(5, truncate=False)
```

```python
print(df.columns)
```

```python
from pyspark.sql.functions import col

df.filter(
    col("Profit").contains("Executive")
).show(truncate=False)
```

# **View Dataset**

```python
df.show(5)
df.printSchema()
df.count()
```

# **Selecting required columns**

```python
selected_df = df.select(
    "Order ID",
    "Category",
    "Region",
    "Sales",
    "Profit"
)
selected_df.show()
```

# **Filtering Electronics Equivalent**

```python
technology_df=df.filter(col("Category") == "Technology")
technology_df.show()
```

# **Filter West Region**

```python
west_df = df.filter(col("Region") == "West")
west_df.show()
```

# **Renaming a column**

```python
df = df.withColumnRenamed("Sales", "Total_Sales")

df.show(5)
```

# **Data type casting**

```python
df = df.withColumn(
    "Total_sales",
    col("Total_sales").cast("double")
)
```

# **Adding new column (18% GST)**

```python
df = df.withColumn(
    "Final_Price",
    col("Total_Sales") * 1.18
)
df.select(
    "Total_Sales",
    "Final_Price"
).show()
```

# **Handling null values"**

```python
df = df.na.fill(0)
```

# **Removing duplicate records**

```python
df = df.dropDuplicates()
```

# **Group by Category**

```python
df.groupBy("Category") \
  .agg(
      sum("Profit").alias("Total Profit")
  ) \
  .show()
```

# **Sorting the data**

```python
df.orderBy(col("Profit").desc()).show()
```

# **Data pipeline**

```python
pipeline = spark.read.csv(
    "Sample - Superstore.csv",
    header=True,
    inferSchema=True
)

pipeline = pipeline.filter(col("Profit") > 500)

pipeline = pipeline.select(
    "Order ID",
    "Category",
    "Sales",
    "Profit"
)

pipeline.show()
```

# **Saving as a CSV**

```python
pipeline.write \
    .mode("overwrite") \
    .option("header", True) \
    .csv("Filtered_Superstore")
```

# **Saving as Parquet**

```python
pipeline.write \
    .mode("overwrite") \
    .parquet("Superstore_Parquet")
```

# **Stop Spark**

```python
spark.stop()
```
