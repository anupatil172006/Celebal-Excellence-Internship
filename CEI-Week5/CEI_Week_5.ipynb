# **Data Cleaning**

**Checking for duplicate rows**

```python
print("Rows before",df.count())
df=df.dropDuplicates()
print("Rows after:",df.count())
```

**Checking and Handling the null values**

```python
df.select([
    count(when(col(c).isNull(),c)).alias(c)
    for c in df.columns
])
```

```python
df=df.na.fill({"Profit":0})
```

```python
df=df.na.fill({"City":"Unknown"})
```

**Schema Modification**

```python
df=df.withColumn("Sales",col("Sales").cast("double"))
df=df.withColumn("Quantity",col("Quantity").cast("int"))
df=df.withColumn("Discount",col("Discount").cast("double"))
```

```python
df.printSchema()
```

# **Filtering**

**Filter furniture Category**

```python
furniture_df=df.filter(col("Category")=="Furniture")
furniture_df.show(5)
```

**Multiple conditions**

```python

df.filter(
    (col("Region")=="West")&
    (col("Category")=="Technology")
).show()
```

# **Aggregation**

**Count**

```python
df.select(count("*").alias("Total Recors")).show()
```

**Sum**

```python
df.select(sum("Profit").alias("Total Profit")).show()
```

**Average**

```python
df.select(avg("Profit").alias("Average Profit")).show()
```

**Minimum**

```python
df.select(min("Profit").alias("Minimum Profit")).show()
```

# **GroupBy**

```python
df.groupBy("Region") \
  .agg(sum("Profit").alias("Total Profit")) \
  .show()
```

```python
df.groupBy("City")\
.count()\
.show()
```

```python
df.groupBy("City")\
.count()\
.filter(col("count")>100)\
.show()
```

# **Renaming a column**

```python
df=df.withColumnRenamed("Customer Name","Customer_Name")
```

# **Converting date column**

```python
from pyspark.sql.functions import to_date

df = df.withColumn(
    "Order_Date",
    to_date(col("Order Date"), "M/d/yyyy")
)
```

# **Final Processing Pipeline**

```python
import pandas as pd

pdf = pd.read_csv("/content/Sample - Superstore.csv", encoding="latin1")

df = spark.createDataFrame(pdf)
```

```python
from pyspark.sql.functions import *

result = (
    df
    .dropDuplicates()
    .na.fill({"City": "Unknown"})
    .groupBy("Region")
    .agg(
        sum("Profit").alias("Total_Profit"),
        count("*").alias("Total_Orders")
    )
)

result.show()
```
